#### Cell 1: Import libraries and load Excel sheets

```python
import pandas as pd
import numpy as np

file_path = "Personal_Loans_Estimation.xlsx"

bank_df = pd.read_excel(file_path, sheet_name="Data")
inflation_df = pd.read_excel(file_path, sheet_name="Inflation rate ")
product_df = pd.read_excel(file_path, sheet_name="Year wise product data")

print("Bank Data:", bank_df.shape)
print("Inflation Data:", inflation_df.shape)
print("Product Data:", product_df.shape)
```

#### Cell 2: Standardize column names

```python
bank_df.columns = bank_df.columns.str.strip()

bank_df.rename(columns={
    "Cunsumer Finance (PKR bn)": "Consumer_Finance_PKR_Bn"
}, inplace=True)

product_df.rename(columns={
    "Auto_Financ": "Auto_Finance"
}, inplace=True)

inflation_df.columns = inflation_df.columns.str.strip()

print(bank_df.columns.tolist())
print(product_df.columns.tolist())
```

#### Cell 3: Check missing values and duplicates

```python
for name, df in {
    "Bank": bank_df,
    "Inflation": inflation_df,
    "Products": product_df
}.items():

    print(f"\n{name} DATA")
    print("Missing Values:\n", df.isnull().sum())
    print("Duplicate Rows:", df.duplicated().sum())
```

#### Cell 4: Validate years and bank records

```python
print("Bank Years:", sorted(bank_df["Year"].unique()))
print("Product Years:", sorted(product_df["Year"].unique()))
print("Inflation Years:", sorted(inflation_df["Year"].unique()))

print("\nTotal Banks:", bank_df["Bank"].nunique())

print("\nBanks per Year:")
print(bank_df.groupby("Year")["Bank"].nunique())

print("\nDuplicate Bank-Year Records:",
      bank_df.duplicated(["Year", "Bank"]).sum())
```

#### Validate product data

```python
product_columns = [
    "Personal_Loans",
    "Home_Finance",
    "Auto_Finance",
    "Credit_Card_Financing"
]

print(product_df[product_columns].describe().round(2))

print("\nNegative Values:")
print((product_df[product_columns] < 0).sum())

print("\nProduct Data by Year:")
print(product_df.to_string(index=False))
```

#### Prepare annual product data

```python
product_analysis = product_df.copy()

product_analysis = product_analysis.sort_values("Year")

product_analysis
```

#### Calculate YoY Growth

```python
products = [
    "Personal_Loans",
    "Auto_Finance",
    "Home_Finance",
    "Credit_Card_Financing"
]

for col in products:
    product_analysis[f"{col}_YoY_%"] = (
        product_analysis[col].pct_change() * 100
    )

product_analysis.round(2)
```

#### Calculate CAGR (2020–2025)

```python
cagr_results = {}

for col in products:
    start_value = product_analysis.iloc[0][col]
    end_value = product_analysis.iloc[-1][col]

    cagr = ((end_value / start_value) ** (1/5) - 1) * 100

    cagr_results[col] = round(cagr, 2)

cagr_df = pd.DataFrame(
    list(cagr_results.items()),
    columns=["Product", "CAGR_%"]
)

cagr_df.sort_values("CAGR_%", ascending=False)
```

Vehicle financing in Pakistan declined sharply in 2022 due to SBP regulatory restrictions, rising interest rates, and soaring car prices. Shorter loan tenures, higher monthly installments, increased down payments, and reduced purchasing power made auto financing unaffordable for many consumers, resulting in falling demand and production cuts.

• Late 2024 (Low Point & Stabilization):
	
    • In December 2024, total auto financing stood at around Rs 235.45 billion, down 6.3% year-on-year compared to December 2023 (Rs 251.25 billion), constrained by high policy rates (which peaked near 22%), steep car prices, and strict regulatory limits.

• 2025 (The Growth Turnaround):
	
    • As the State Bank of Pakistan (SBP) aggressively cut interest rates throughout 2025 (lowering the policy rate from 22% down toward 11%), consumer car loans entered a prolonged streak of consecutive monthly growth.

#### Latest year bank performance (2025)

```python
latest = bank_df[bank_df["Year"] == 2025].copy()

latest = latest.sort_values(
    "Consumer_Finance_PKR_Bn",
    ascending=False
)

latest[[
    "Bank",
    "Consumer_Finance_PKR_Bn",
    "Advances (PKR Bn)",
    "Total Assets (PKR Bn)",
    "Total Deposits (PKR Bn)"
]].head(15)
```

#### Calculate consumer finance share of advances

```python
latest["Consumer_Finance_Share_%"] = (
    latest["Consumer_Finance_PKR_Bn"] /
    latest["Advances (PKR Bn)"]
) * 100

latest.sort_values(
    "Consumer_Finance_Share_%",
    ascending=False
)[["Bank", "Consumer_Finance_Share_%"]].head(15)
```

#### Calculate bank-wise YoY growth

```python
bank_analysis = bank_df.sort_values(
    ["Bank", "Year"]
).copy()

bank_analysis["YoY_Growth_%"] = (
    bank_analysis.groupby("Bank")["Consumer_Finance_PKR_Bn"]
    .pct_change() * 100
)

bank_analysis[
    bank_analysis["Year"] == 2025
][["Bank", "Consumer_Finance_PKR_Bn", "YoY_Growth_%"]] \
.sort_values("YoY_Growth_%", ascending=False)
```

#### Top 10 banks visualization

```python
import plotly.express as px

top10 = latest.nlargest(10, "Consumer_Finance_PKR_Bn")

fig = px.bar(
    top10,
    x="Consumer_Finance_PKR_Bn",
    y="Bank",
    orientation="h",
    text_auto=".1f",
    title="Top 10 Banks by Consumer Finance (2025)",
    labels={
        "Consumer_Finance_PKR_Bn": "Consumer Finance (PKR Bn)",
        "Bank": ""
    },
    color_discrete_sequence=["#404040"]
)

fig.update_layout(yaxis={"categoryorder": "total ascending"})

fig.show()
```

#### Market share of top banks

```python
latest["Market_Share_%"] = (
    latest["Consumer_Finance_PKR_Bn"] /
    latest["Consumer_Finance_PKR_Bn"].sum()
) * 100

top10_share = latest.nlargest(10, "Market_Share_%")

fig = px.pie(
    top10_share,
    names="Bank",
    values="Market_Share_%",
    title="Top 10 Banks — Consumer Finance Market Share (2025)",
    color_discrete_sequence=[
        "#000000",
        "#404040",
        "#737373",
        "#A6A6A6",
        "#D9D9D9",
        "#595959",
        "#BFBFBF",
        "#262626",
        "#E5E5E5",
        "#8C8C8C"
    ]
)


fig.update_layout(
    width=800,
    height=500
)
fig.update_traces(
    textinfo="percent+label",
    textposition="inside",
    marker=dict(line=dict(width=0))
)

fig.show()
```

#### Calculate bank-wise YoY growth

```python
# Calculate market share of all banks
latest["Market_Share_%"] = (
    latest["Consumer_Finance_PKR_Bn"] /
    latest["Consumer_Finance_PKR_Bn"].sum()
) * 100

# Calculate concentration of top banks
top_5_share = latest.nlargest(5, "Consumer_Finance_PKR_Bn")["Market_Share_%"].sum()
top_10_share = latest.nlargest(10, "Consumer_Finance_PKR_Bn")["Market_Share_%"].sum()

other_banks_share = 100 - top_10_share

concentration = pd.DataFrame({
    "Category": ["Top 5 Banks", "Top 10 Banks", "Other Banks"],
    "Market Share (%)": [
        top_5_share,
        top_10_share,
        other_banks_share
    ]
})

fig = px.bar(
    concentration,
    x="Category",
    y="Market Share (%)",
    text_auto=".2f",
    title="Consumer Finance Market Concentration — 2025",
    color="Category",
    color_discrete_sequence=["#000000", "#737373", "#BFBFBF"]
)


fig.update_layout(
    width=600,
    height=500,
    showlegend=False,
    yaxis_title="Market Share (%)",
    xaxis_title=""
)

fig.update_traces(
    textposition="outside"
)

fig.show()

print(f"Top 5 Banks Market Share: {top_5_share:.2f}%")
print(f"Top 10 Banks Market Share: {top_10_share:.2f}%")
print(f"Other Banks Market Share: {other_banks_share:.2f}%")
```

```python
import plotly.express as px

# Table
inflation_table = inflation_df.copy()
inflation_table["Inflation_%"] = (inflation_table["Avg_Inflation_Rate"] * 100).round(2)
display(inflation_table[["Year", "Inflation_%"]])

# Chart
fig = px.bar(
    inflation_table,
    x="Year",
    y="Inflation_%",
    text_auto=".1f",
    title="Pakistan Average Inflation Rate (2020–2025)",
    labels={"Inflation_%": "Inflation (%)", "Year": "Year"},
    color_discrete_sequence=["#404040"]
)

fig.update_traces(textposition="outside")
fig.update_layout(width=800, height=40)
fig.show()
```

```python
# Step 1: Price index (2020 = 1)
inflation = inflation_df.set_index("Year")["Avg_Inflation_Rate"]
price_index = (1 + inflation).cumprod()
price_index = price_index / price_index.loc[2020]

# Step 2: Real values (2020 prices)
real_df = product_df.set_index("Year")[products].div(price_index, axis=0)

# Step 3: Nominal vs Real CAGR
nominal_cagr = (product_df.set_index("Year")[products].iloc[-1] /
                product_df.set_index("Year")[products].iloc[0]) ** (1/5) - 1
real_cagr = (real_df.iloc[-1] / real_df.iloc[0]) ** (1/5) - 1

result = pd.DataFrame({
    "Nominal_CAGR_%": nominal_cagr * 100,
    "Real_CAGR_%": real_cagr * 100
}).round(1).sort_values("Real_CAGR_%", ascending=False)

display(real_df.round(1))
display(result)
```
