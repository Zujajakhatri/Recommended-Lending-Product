# Consumer Finance in Pakistan – Data Processing & Analysis

### Preparing and Analysing Bank-Level Consumer Finance Data for DRB

**Prepared by:** Zujaja  
**Tools Used:** Python, Pandas, NumPy, Plotly

## Project Overview

This project supports a DRB report on **Consumer Finance in Pakistan**, focusing on three lending products: **Personal Loans**, **Auto / Vehicle Finance** and **Housing / Home Finance**.

Bank-level data for these three products is not publicly available. This notebook therefore analyses **overall Consumer Finance**, which forms the quantitative foundation of the report. The three products are covered in the report through qualitative, research-based analysis.

## Dataset Note

This project uses a custom-built dataset developed for analytical purposes.

- **Consumer Finance (PKR Bn)** is estimated as **Advances × Consumer Finance share of advances**.
- The dataset covers **39 banks** (commercial, Islamic, foreign, microfinance and digital banks) for **2020–2025**.
- The data represents **overall consumer finance only** and is **not** split into personal loans, auto finance or housing finance.
- All Excel values are used as they are.

## Workflow

1. Import libraries and load the dataset
2. Inspect data structure and quality
3. Clean the data (remove duplicates, rename columns, flag missing data)
4. Review data quality checks
5. Create new features
6. Build a yearly market summary and calculate CAGR
7. Export final datasets
8. Visualize market-level trends
9. Visualize bank-level analysis

# Part 1: Data Preparation

#### Import Required Libraries

Pandas and NumPy are used for data handling and calculations.

```python
# Data Handling
import pandas as pd
import numpy as np

# Warnings
import warnings
warnings.filterwarnings("ignore")
```

#### Outcome

All required libraries were imported successfully.

#### Load the Dataset

The **Data** sheet of the Excel workbook is loaded as the main dataset.

```python
FILE = "Personal_Loans_Estimation.xlsx"

df = pd.read_excel(FILE, sheet_name="Data")

df.head()
```

#### Outcome

The dataset was loaded successfully. The first five rows are displayed above.

#### Inspect Dataset Structure and Data Quality

The dataset is checked for its size, duplicate rows, missing values and data types.

```python
print("Rows and columns:", df.shape)
print("-" * 35)
print("Duplicate Rows:", df.duplicated().sum())
print("-" * 35)
print("Null Values:\n", df.isnull().sum())
print("-" * 35)
print("Data Types:\n", df.dtypes)
```

#### Outcome

- The dataset has **234 rows and 13 columns** (39 banks × 6 years).
- There are **no duplicate rows**.
- **39 missing values** appear in opening balance, change in balance and average balance. These are the **2020 rows**, because 2020 is the first year and has no previous balance.
- All data types are correct.

#### Outcome

Both columns were confirmed as exact copies and removed. The dataset now has **11 columns**.

```python
print("Total Assets Year End = Total Assets:",
      df["Total Assets Year End (PKR Bn)"].equals(df["Total Assets (PKR Bn)"]))
print("Total Deposits (repeat) = Total Deposits:",
      df["Total Deposits (PKR Bn).1"].equals(df["Total Deposits (PKR Bn)"]))

df = df.drop(columns=["Total Assets Year End (PKR Bn)", "Total Deposits (PKR Bn).1"])

print("Columns after cleaning:", df.shape[1])
```

#### Outcome

Both columns were confirmed as exact copies and removed. The dataset now has **13 columns**.

#### Rename Columns

Column names are shortened and standardized for easier use. Values are not changed.

```python
df = df.rename(columns={
    "Bank Type": "Bank_Type",
    "Advances (PKR Bn)": "Advances_PKR_Bn",
    "Total Assets (PKR Bn)": "Total_Assets_PKR_Bn",
    "Total Deposits (PKR Bn)": "Total_Deposits_PKR_Bn",
    "Consumer Finance  share of advances (%)": "CF_Share_of_Advances",
    "Consumer Finance (PKR Bn)": "Consumer_Finance_PKR_Bn",
    "Opening Consumer Finance (PKR Bn)": "Opening_CF_PKR_Bn",
    "Est. Disbursement in Year (PKR Bn)": "Est_Disbursement_PKR_Bn",
    "Auto check": "Auto_Check",
    "Change in balance (helper)": "Change_in_CF_PKR_Bn",
    "Average balance (helper)": "Average_CF_PKR_Bn",
})

df["Year"] = df["Year"].astype(int)
df["Bank"] = df["Bank"].str.strip()
df["Bank_Type"] = df["Bank_Type"].str.strip()

df.head()
```

#### Outcome

All columns were renamed successfully and **Year** is stored as an integer.

#### Identify Banks Without Consumer Finance Data

Some banks show a 0% consumer finance share in all years, which means data was not available. These banks are marked in a new column, **CF_Data_Available**.

```python
no_data = df.groupby("Bank")["CF_Share_of_Advances"].sum()
no_data_banks = no_data[no_data == 0].index

print("Banks with no consumer finance data:", len(no_data_banks))
for b in no_data_banks:
    print(" -", b)

df["CF_Data_Available"] = np.where(df["Bank"].isin(no_data_banks), "No", "Yes")
```

#### Outcome

**10 banks** have no consumer finance data, mainly foreign banks, new digital banks and some microfinance banks. They are marked **"No"** and excluded from the analysis.

**29 banks** have consumer finance data and are used in the analysis.

#### Check Negative Estimated Disbursement

Estimated disbursement is checked for negative values, which are not possible in reality.

# Part 2: Feature Engineering

#### Create New Features

Three new features are added:
- **CF_YoY_Growth** – yearly growth of each bank's consumer finance
- **CF_to_Assets** – consumer finance as a share of total assets
- **Advances_to_Deposits** – share of deposits that the bank has lent out

```python
df = df.sort_values(["Bank", "Year"]).reset_index(drop=True)

# Yearly growth of each bank's consumer finance
df["CF_YoY_Growth"] = (
    df.groupby("Bank")["Consumer_Finance_PKR_Bn"].pct_change(fill_method=None)
).replace([np.inf, -np.inf], np.nan).round(4)

# Consumer finance as share of total assets
df["CF_to_Assets"] = (
    df["Consumer_Finance_PKR_Bn"] / df["Total_Assets_PKR_Bn"]
).replace([np.inf, -np.inf], np.nan).round(4)

# Advances to deposits ratio
df["Advances_to_Deposits"] = (
    df["Advances_PKR_Bn"] / df["Total_Deposits_PKR_Bn"]
).replace([np.inf, -np.inf], np.nan).round(4)

df[["Year", "Bank", "Consumer_Finance_PKR_Bn", "CF_YoY_Growth",
    "CF_Share_of_Advances", "CF_to_Assets", "Advances_to_Deposits"]].head(6)
```

#### Outcome

All three features were added successfully. Growth is blank for 2020 (first year) and for banks without consumer finance data.

```python
cf = df[df["CF_Data_Available"] == "Yes"]

summary = cf.groupby("Year").agg(
    Banks_with_Data=("Bank", "nunique"),
    Total_Advances_PKR_Bn=("Advances_PKR_Bn", "sum"),
    Consumer_Finance_PKR_Bn=("Consumer_Finance_PKR_Bn", "sum"),
    Banks_Growing=("Change_in_CF_PKR_Bn", lambda s: (s > 0).sum()),
    Banks_Declining=("Change_in_CF_PKR_Bn", lambda s: (s < 0).sum()),
).reset_index()

summary["CF_YoY_Growth"] = summary["Consumer_Finance_PKR_Bn"].pct_change().round(4)
summary["CF_Penetration"] = (
    summary["Consumer_Finance_PKR_Bn"] / summary["Total_Advances_PKR_Bn"]
).round(4)

summary.loc[summary["Year"] == summary["Year"].min(),
            ["Banks_Growing", "Banks_Declining"]] = np.nan

summary
```

# Part 3: Consumer Finance Market Summary

#### Create Yearly Summary

Bank-level data is combined into yearly market totals. Only the **29 banks with consumer finance data** are included.

#### Calculate CAGR

CAGR shows the average yearly growth of the consumer finance market from 2020 to 2025.

```python
start = summary.loc[summary["Year"] == 2020, "Consumer_Finance_PKR_Bn"].values[0]
end   = summary.loc[summary["Year"] == 2025, "Consumer_Finance_PKR_Bn"].values[0]

cagr = (end / start) ** (1 / 5) - 1

print(f"Consumer Finance 2020: PKR {start:,.1f} bn")
print(f"Consumer Finance 2025: PKR {end:,.1f} bn")
print(f"CAGR 2020–2025: {cagr:.1%}")
```

#### Outcome

The consumer finance market grew at a **CAGR of 17.4%**, more than doubling from PKR 672.6bn to PKR 1,499.7bn.



#### Select Final Features

The required columns are selected to create the final dataset (`df_final`).

```python
df_final = df[
    [
        "Year",
        "Bank",
        "Bank_Type",
        "Advances_PKR_Bn",
        "Total_Assets_PKR_Bn",
        "Total_Deposits_PKR_Bn",
        "CF_Data_Available",
        "CF_Share_of_Advances",
        "Consumer_Finance_PKR_Bn",
        "Opening_CF_PKR_Bn",
        "Change_in_CF_PKR_Bn",
        "Average_CF_PKR_Bn",
        "CF_YoY_Growth",
        "CF_to_Assets",
        
    ]
].copy()

df_final.info()
```

#### Outcome

The final dataset has **234 rows and 14 columns**. Missing values are expected for 2020 (first year), for banks without consumer finance data, and for 2 new banks (Mashreq and Raqami) that report zero assets.

#### Export Final Datasets

The bank-level dataset and the yearly summary are saved as CSV files.

```python
df_final.to_csv("a. Consumer_Finance_Bank_Level.csv", index=False, encoding="utf-8-sig")
summary.to_csv("a. Consumer_Finance_Yearly_Summary.csv", index=False, encoding="utf-8-sig")

print("CSV files saved successfully!")
print("Bank level:", df_final.shape)
print("Yearly summary:", summary.shape)
```

#### Outcome

Both files were saved successfully:
- **Bank level:** 234 rows × 14 columns
- **Yearly summary:** 6 rows × 8 columns

# Part 4: Market-Level Visualizations

#### Chart Setup

Plotly is used for all charts. One simple theme (white background with black and grey colors) is applied to all visuals.

```python
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import os

os.makedirs("charts", exist_ok=True)

LAYOUT = dict(
    template="plotly_white",
    font=dict(family="Arial", size=13),
    title_font=dict(size=18),
    title_x=0.5,
    height=500,
    width=900,
    margin=dict(t=80, l=70, r=40, b=70),
)

def save(fig, name):
    fig.write_html(f"charts/{name}.html")
    try:
        fig.write_image(f"charts/{name}.png", scale=2)
    except Exception:
        print("PNG not saved (needs kaleido + Google Chrome). HTML saved.")
    fig.show()

summary["Year_Label"] = summary["Year"].astype(str)
```

```python
import plotly.express as px

# One theme for all charts: white background, black and grey colors
px.defaults.template = "simple_white"
px.defaults.color_discrete_sequence = ["#000000", "#808080", "#c0c0c0", "#404040"]

# Copy of summary for charts (Year as text so it shows properly on the axis)
chart = summary.copy()
chart["Year"] = chart["Year"].astype(str)
```

### Consumer Finance Trend (2020–2025)

```python
fig = px.bar(chart, x="Year", y="Consumer_Finance_PKR_Bn", text_auto=".0f",
             title="Consumer Finance in Pakistan (2020–2025)",
             labels={"Consumer_Finance_PKR_Bn": "PKR Billion"})


fig.show()
```

#### Outcome

The consumer finance market more than **doubled** in five years, from PKR 673bn to PKR 1,500bn. The biggest increase came in 2025.

### Year-on-Year Growth

```python
total = df.groupby("Year")["Consumer_Finance_PKR_Bn"].sum()

for year, value in total.items():
    print(f"{year}: {value:,.0f}")
```

```python
growth = chart.dropna(subset=["CF_YoY_Growth"]).copy()
growth["Trend"] = np.where(growth["CF_YoY_Growth"] >= 0, "Growth", "Decline")

fig = px.bar(growth, x="Year", y="CF_YoY_Growth", color="Trend", text_auto=".1%",
             title="Consumer Finance – Year-on-Year Growth",
             labels={"CF_YoY_Growth": "Growth"},
             color_discrete_map={"Growth": "#000000", "Decline": "#a0a0a0"})

fig.update_layout(yaxis_tickformat=".0%")
fig.show()
```

#### Outcome

The market grew every year. Growth was strongest in **2021 (33.8%)** and **2025 (28.3%)**, and slowed sharply in **2023 (8.5%)** and **2024 (4.6%)**. The reasons for the slowdown will be explained through external research (e.g. interest rates and SBP regulations).

```python
fig = px.line(chart, x="Year", y="CF_Penetration", markers=True, text="CF_Penetration",
              title="Consumer Finance as % of Total Advances",
              labels={"CF_Penetration": "Share of Advances"})

fig.update_traces(texttemplate="%{y:.1%}", textposition="top center")
fig.update_layout(yaxis_tickformat=".0%", yaxis_range=[0, 0.12])
fig.show()
```

```python
banks = chart.dropna(subset=["Banks_Growing"])
banks = banks.rename(columns={"Banks_Growing": "Growing", "Banks_Declining": "Declining"})

fig = px.bar(banks, x="Year", y=["Growing", "Declining"], barmode="group", text_auto=True,
             title="Number of Banks with Growing vs Declining Consumer Finance",
             labels={"value": "Number of Banks", "variable": ""})
fig.show()
```

```python
mix = cf.groupby(["Year", "Bank_Type"], as_index=False)["Consumer_Finance_PKR_Bn"].sum()
mix["Year"] = mix["Year"].astype(str)

fig = px.bar(mix, x="Year", y="Consumer_Finance_PKR_Bn", color="Bank_Type",
             title="Consumer Finance Mix by Bank Type",
             labels={"Consumer_Finance_PKR_Bn": "Share of Consumer Finance (%)",
                     "Bank_Type": "Bank Type"})

fig.update_layout(barnorm="percent")
fig.show()
```

```python
def short_name(name):
    name = name.replace("(Pakistan)", "").strip()
    if "(" in name:
        return name.split("(")[-1].replace(")", "")
    for word in [" Limited", " Pakistan", " Bank", " N.A."]:
        name = name.replace(word, "")
    return name.replace(" Microfinance", " MFB").strip()

cf = cf.copy()
cf["Bank_Short"] = cf["Bank"].apply(short_name)

# 2025 data with market share
latest = cf[cf["Year"] == 2025].copy()
latest["Market_Share"] = latest["Consumer_Finance_PKR_Bn"] / latest["Consumer_Finance_PKR_Bn"].sum()

cf["Bank_Short"].unique()
```

```python
avg_cf = cf.groupby("Bank_Short", as_index=False)["Consumer_Finance_PKR_Bn"].mean()

top10_avg = avg_cf.sort_values("Consumer_Finance_PKR_Bn", ascending=False).head(10)

fig = px.bar(top10_avg, x="Consumer_Finance_PKR_Bn", y="Bank_Short", orientation="h", text_auto=".0f",
             title="Top 10 Banks by Average Consumer Finance (2020–2025)",
             labels={"Consumer_Finance_PKR_Bn": "Average PKR Billion", "Bank_Short": ""})

fig.update_layout(yaxis={"categoryorder": "total ascending"})
fig.show()
```

```python
share = cf[cf["Year"].isin([2020, 2025])].copy()
share["Market_Share"] = share["Consumer_Finance_PKR_Bn"] / share.groupby("Year")["Consumer_Finance_PKR_Bn"].transform("sum")
share["Year"] = share["Year"].astype(str)
share = share[share["Bank_Short"].isin(top10_avg["Bank_Short"])]

fig = px.bar(share, x="Bank_Short", y="Market_Share", color="Year", barmode="group", text_auto=".0%",
             title="Market Share of Top 10 Banks: 2020 vs 2025",
             labels={"Market_Share": "Market Share", "Bank_Short": ""},
             category_orders={"Bank_Short": top10_avg["Bank_Short"].tolist()},
             color_discrete_map={"2020": "#c0c0c0", "2025": "#000000"})

fig.update_layout(yaxis_tickformat=".0%")
fig.show()
```

```python
first = cf[cf["Year"] == 2020].set_index("Bank_Short")["Consumer_Finance_PKR_Bn"]
last  = cf[cf["Year"] == 2025].set_index("Bank_Short")["Consumer_Finance_PKR_Bn"]

cagr_bank = ((last / first) ** (1 / 5) - 1).reset_index(name="CAGR")
cagr_bank["Trend"] = np.where(cagr_bank["CAGR"] >= 0, "Growth", "Decline")

fig = px.bar(cagr_bank, x="CAGR", y="Bank_Short", orientation="h", color="Trend", text_auto=".1%",
             title="Consumer Finance Growth by Bank – CAGR 2020–2025",
             labels={"CAGR": "CAGR", "Bank_Short": ""},
             color_discrete_map={"Growth": "#000000", "Decline": "#a0a0a0"}, height=750)

fig.update_layout(yaxis={"categoryorder": "total ascending"}, xaxis_tickformat=".0%")
fig.show()
```

```python
focus = latest.sort_values("CF_Share_of_Advances", ascending=False).head(15)

fig = px.bar(focus, x="CF_Share_of_Advances", y="Bank_Short", orientation="h", text_auto=".1%",
             title="Retail Focus: Consumer Finance as % of Each Bank's Advances (2025)",
             labels={"CF_Share_of_Advances": "Share of Advances", "Bank_Short": ""}, height=600)

fig.update_layout(yaxis={"categoryorder": "total ascending"}, xaxis_tickformat=".0%")
fig.show()
```
