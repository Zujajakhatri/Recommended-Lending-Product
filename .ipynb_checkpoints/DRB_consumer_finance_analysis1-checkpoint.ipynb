# Consumer Finance Analysis – DRB
Yeh notebook aapki source Excel file (Year, Bank, Bank Type, Advances, Assets, Deposits, Consumer Finance) se poori analysis file banati hai: 15 sheets, wahi structure jo `DRB_consumer_finance_analysis.xlsx` mein hai.

**Kaise chalayein:** Sirf **Step 1** mein file ka path check karein, phir menu se **Run All** karein (Jupyter: *Kernel → Restart & Run All*, Colab: *Runtime → Run all*). Naya data aaye to source file update karein aur dobara Run All karein.

Zaroori libraries: `pandas`, `numpy`, `openpyxl`. Agar nahi hain to: `pip install pandas numpy openpyxl`

## Step 1 – Settings (sirf yahan cheezein badlein)

```python
import pandas as pd
import numpy as np

# ---- Files ----
INPUT_FILE  = "Personal_Loans_Estimation.xlsx"          # source data
INPUT_SHEET = "Data"
OUTPUT_FILE = "DRB_consumer_finance_analysis_new.xlsx"  # yeh file banegi

# Source file ke pehle 7 columns is order mein hone chahiye:
# Year | Bank | Bank Type | Advances | Total Assets | Total Deposits | Consumer Finance (PKR bn)
COLS = ["year", "bank", "bank_type", "advances", "assets", "deposits", "cf_excel"]

# ---- Disbursement model: loan tenure (saal) ----
TENURES = {"low": 4.0, "base": 3.0, "high": 2.0}

# ---- Issue check thresholds ----
CF_CHANGE_FLAG_PCT     = 50   # CF ek saal mein itne % se zyada badle to flag
ASSETS_CHANGE_FLAG_PCT = 80   # Assets ek saal mein itne % se zyada badlein to flag

# ---- Ex-HBL total ke liye ----
EXCLUDE_BANK = "Habib Bank Limited (HBL)"

# ---- DRB peer banks ----
PEERS = [
    "Standard Chartered Bank (Pakistan) Limited", "Dubai Islamic Bank Pakistan Limited",
    "Al Baraka Bank (Pakistan) Limited", "JS Bank Limited", "BankIslami Pakistan Limited",
    "Faysal Bank Limited", "Soneri Bank Limited", "Habib Metropolitan Bank Limited",
    "Bank of Khyber (BOK)", "Sindh Bank Limited",
]
```

## Step 3 – Data load aur clean

```python
raw = pd.read_excel(INPUT_FILE, sheet_name=INPUT_SHEET, usecols="A:G")
raw.columns = COLS
raw = raw.dropna(subset=["year", "bank"])
raw["year"] = pd.to_numeric(raw["year"], errors="coerce").astype(int)   # kuch years text mein hote hain
raw["bank"] = raw["bank"].astype(str).str.strip()
raw["bank_type"] = raw["bank_type"].astype(str).str.strip()
for c in ["advances", "assets", "deposits", "cf_excel"]:
    raw[c] = pd.to_numeric(raw[c], errors="coerce").fillna(0)

# Bank group (missing hon to Bank Type se)
cd = raw.copy()
cd["bank_group"] = cd["bank"].map(BANK_GROUP)
missing = cd.loc[cd["bank_group"].isna(), "bank"].unique()
if len(missing):
    print("WARNING – yeh banks BANK_GROUP mein nahi hain, Bank Type se group lagaya:", list(missing))
    cd.loc[cd["bank_group"].isna(), "bank_group"] = cd["bank_type"].map(TYPE_TO_GROUP).fillna("Other")

# CF: 0 = data available nahi
cd["cf_bn"] = cd["cf_excel"].where(cd["cf_excel"] > 0)
cd["cf_share"] = cd["cf_bn"] / cd["advances"].replace(0, np.nan)
cd["review_note"] = [REVIEW_NOTES.get((y, b)) for y, b in zip(cd["year"], cd["bank"])]

cd = cd.sort_values(["bank", "year"]).reset_index(drop=True)
cd["cf_opening"] = cd.groupby("bank")["cf_bn"].shift(1)
cd["cf_change"]  = cd["cf_bn"] - cd["cf_opening"]
cd["cf_avg"]     = (cd["cf_bn"] + cd["cf_opening"]) / 2
for k, t in TENURES.items():
    cd["disb_" + k] = cd["cf_change"] + cd["cf_avg"] / t      # MODELLED ESTIMATE

clean_data = cd[["year", "bank", "bank_type", "bank_group", "advances", "assets", "deposits", "cf_share",
                 "cf_bn", "cf_excel", "cf_opening", "cf_change", "cf_avg", "disb_low", "disb_base",
                 "disb_high", "review_note"]]

Y0, Y1 = cd["year"].min(), cd["year"].max()
N_YEARS = Y1 - Y0
print(f"{cd['bank'].nunique()} banks, years {Y0}-{Y1}")
```

## Step 4 – Bank-level metrics (har bank, har saal)

```python
def size_tier(a):
    if a >= 3000: return "Large (≥ 3,000bn assets)"
    if a >= 1000: return "Medium (1,000–3,000bn)"
    if a >= 200:  return "Small (200–1,000bn)"
    return "Very small (< 200bn)"

has_data = cd.groupby("bank")["cf_bn"].apply(lambda s: s.notna().any())
BANKS = has_data[has_data].index
print("Banks with CF data:", len(BANKS), "| without:", list(has_data[~has_data].index))

b = cd[cd["bank"].isin(BANKS)].copy()
b["short"] = b["bank"].map(SHORT).fillna(b["bank"])
b["yoy_growth_pct"]   = b["cf_change"] / b["cf_opening"] * 100
b["abs_change_bn"]    = b["cf_change"]
b["market_share_pct"] = b["cf_bn"] / b.groupby("year")["cf_bn"].transform("sum") * 100
b["rank"]             = b.groupby("year")["cf_bn"].rank(ascending=False, method="min").astype("Int64")
b["cf_share_pct"]     = b["cf_share"] * 100
b["cf_pct_of_assets"] = b["cf_bn"] / b["assets"].replace(0, np.nan) * 100
b["adr_pct"]          = b["advances"] / b["deposits"].replace(0, np.nan) * 100
b["size_tier"]        = b["assets"].apply(size_tier)

bank_metrics_all_years = b[["year", "bank", "bank_type", "advances", "assets", "deposits", "cf_share", "cf_excel",
    "review_note", "cf_bn", "bank_group", "short", "cf_opening", "cf_change", "cf_avg", "disb_base", "disb_high",
    "disb_low", "yoy_growth_pct", "abs_change_bn", "market_share_pct", "rank", "cf_share_pct", "cf_pct_of_assets",
    "adr_pct", "size_tier"]].reset_index(drop=True)

# CAGR (pehle saal se aakhri saal)
piv = b.pivot(index="bank", columns="year", values="cf_bn")
cagr = ((piv[Y1] / piv[Y0]) ** (1 / N_YEARS) - 1) * 100

last = b[b["year"] == Y1].copy()
last["cagr_pct"] = last["bank"].map(cagr)
bank_metrics_last = last.sort_values("rank")[["bank", "short", "bank_group", "size_tier", "assets", "advances",
    "deposits", "cf_bn", "cf_share_pct", "market_share_pct", "rank", "yoy_growth_pct", "cf_pct_of_assets",
    "adr_pct", "cagr_pct"]].reset_index(drop=True)
bank_metrics_last.head(10)
```

## Step 5 – Market summary aur concentration

```python
rows = []
for y, d in b.groupby("year"):
    d = d[d["cf_bn"].notna()]
    r = dict(year=y, banks_with_data=len(d), consumer_finance_bn=d["cf_bn"].sum(),
             advances_bn=d["advances"].sum(), deposits_bn=d["deposits"].sum(),
             disb_low_bn=d["disb_low"].sum(min_count=1), disb_base_bn=d["disb_base"].sum(min_count=1),
             disb_high_bn=d["disb_high"].sum(min_count=1))
    r["penetration_pct"] = r["consumer_finance_bn"] / r["advances_bn"] * 100
    r["ex_hbl_bn"] = d.loc[d["bank"] != EXCLUDE_BANK, "cf_bn"].sum()
    r["banks_growing"]   = (d["yoy_growth_pct"] > 0).sum() if y > Y0 else np.nan
    r["banks_declining"] = (d["yoy_growth_pct"] < 0).sum() if y > Y0 else np.nan
    rows.append(r)
market_summary = pd.DataFrame(rows)
market_summary["yoy_growth_pct"] = market_summary["consumer_finance_bn"].pct_change() * 100
market_summary["ex_hbl_yoy_pct"] = market_summary["ex_hbl_bn"].pct_change() * 100
market_summary = market_summary[["year", "banks_with_data", "consumer_finance_bn", "advances_bn", "deposits_bn",
    "disb_low_bn", "disb_base_bn", "disb_high_bn", "yoy_growth_pct", "penetration_pct", "ex_hbl_bn",
    "ex_hbl_yoy_pct", "banks_growing", "banks_declining"]]

rows = []
for y, d in b.groupby("year"):
    s = d["market_share_pct"].dropna().sort_values(ascending=False)
    rows.append(dict(year=y, top3_share_pct=s.head(3).sum(), top5_share_pct=s.head(5).sum(),
                     top10_share_pct=s.head(10).sum(), hhi=(s ** 2).sum(),
                     largest_bank=d.loc[d["cf_bn"].idxmax(), "bank"]))
concentration = pd.DataFrame(rows)
market_summary
```

## Step 6 – Contribution to growth aur bank groups

```python
contribution_to_growth = pd.DataFrame({"bank": piv.index, "cf_start_bn": piv[Y0].values, "cf_end_bn": piv[Y1].values})
contribution_to_growth["increase_bn"] = contribution_to_growth["cf_end_bn"] - contribution_to_growth["cf_start_bn"]
contribution_to_growth["share_of_total_increase_pct"] = (contribution_to_growth["increase_bn"]
                                                         / contribution_to_growth["increase_bn"].sum() * 100)
contribution_to_growth = contribution_to_growth.sort_values("increase_bn", ascending=False).reset_index(drop=True)

g_cf  = b.pivot_table(index="bank_group", columns="year", values="cf_bn", aggfunc="sum")
g_adv = b[b["cf_bn"].notna()].pivot_table(index="bank_group", columns="year", values="advances", aggfunc="sum")

group_cf_bn = g_cf.copy()
group_cf_bn["cagr_pct"] = ((g_cf[Y1] / g_cf[Y0]) ** (1 / N_YEARS) - 1) * 100
group_share_pct       = g_cf / g_cf.sum() * 100
group_penetration_pct = g_cf / g_adv * 100
group_cf_bn, group_share_pct, group_penetration_pct = [t.reset_index().rename_axis(None, axis=1)
    for t in (group_cf_bn, group_share_pct, group_penetration_pct)]
group_cf_bn
```

## Step 7 – Stability profile aur DRB peers
Profile rules: CAGR < 0 → *Declining*; 2+ saal decline → *Volatile / recovering*; CAGR ≥ 20% → *Consistent strong grower*; baqi → *Moderate grower*

```python
def profile(c, dec):
    if c < 0:    return "Declining"
    if dec >= 2: return "Volatile / recovering"
    if c >= 20:  return "Consistent strong grower"
    return "Moderate grower"

rows = []
for bk, d in b.groupby("bank"):
    yy = d["yoy_growth_pct"].dropna()
    rows.append(dict(bank=bk, years_with_data=int(d["cf_bn"].notna().sum()), avg_yoy_growth_pct=yy.mean(),
                     volatility_pct=yy.std(), years_declined=int((yy < 0).sum()), cagr_pct=cagr[bk]))
stability = pd.DataFrame(rows)
stability["profile"] = [profile(c, d) for c, d in zip(stability["cagr_pct"], stability["years_declined"])]
stability = stability.sort_values("cagr_pct", ascending=False).reset_index(drop=True)

drb_peers = (bank_metrics_last[bank_metrics_last["bank"].isin(PEERS)]
             .sort_values("cf_share_pct", ascending=False).reset_index(drop=True))
missing_peers = set(PEERS) - set(drb_peers["bank"])
if missing_peers: print("WARNING – yeh peers data mein nahi mile:", missing_peers)
drb_peers
```

## Step 8 – Data quality issues

```python
iss = []
for bk in sorted(set(cd["bank"]) - set(BANKS)):
    iss.append(("All", bk, "No consumer finance data", "CF is 0/blank in all years; excluded from consumer finance analysis"))
for r in cd.itertuples():
    if r.advances > r.assets:
        iss.append((str(r.year), r.bank, "Advances > Assets", f"Advances {r.advances:.2f} vs Assets {r.assets:.2f}"))
for r in cd.itertuples():
    if r.assets > 0 and r.deposits > r.assets:
        iss.append((str(r.year), r.bank, "Deposits > Assets", f"Deposits {r.deposits:.2f} vs Assets {r.assets:.2f}"))
for r in cd.itertuples():
    if pd.notna(r.cf_bn) and r.cf_bn > r.advances:
        iss.append((str(r.year), r.bank, "Consumer finance > Advances", f"CF {r.cf_bn:.2f} vs Advances {r.advances:.2f}"))
for r in b.itertuples():
    if pd.notna(r.yoy_growth_pct) and abs(r.yoy_growth_pct) >= CF_CHANGE_FLAG_PCT - 0.5:
        iss.append((str(r.year), r.bank, f"Consumer finance change > {CF_CHANGE_FLAG_PCT}% in one year",
                    f"YoY change {r.yoy_growth_pct:+.0f}%"))
cd["assets_yoy"] = cd.groupby("bank")["assets"].pct_change() * 100
for r in cd.itertuples():
    v = r.assets_yoy
    if pd.notna(v) and np.isfinite(v) and abs(v) > ASSETS_CHANGE_FLAG_PCT:
        iss.append((str(r.year), r.bank, f"Assets change > {ASSETS_CHANGE_FLAG_PCT}% in one year",
                    f"YoY change {v:+.0f}% (check consolidated vs standalone)"))
for r in cd.itertuples():
    if isinstance(r.review_note, str):
        iss.append((str(r.year), r.bank, "Excel review note", r.review_note))
for r in b.itertuples():
    if pd.notna(r.disb_base) and r.disb_base < 0:
        iss.append((str(r.year), r.bank, "Negative modelled disbursement",
                    f"Base-case disbursement {r.disb_base:.2f}; model not valid for this row"))
issues = pd.DataFrame(iss, columns=["year", "bank", "check", "detail"])
issues["check"].value_counts()
```

## Step 9 – Sanity checks (agar koi ghalti ho to yahan error aayega)

```python
assert (bank_metrics_all_years.groupby("year")["market_share_pct"].sum() - 100).abs().max() < 1e-6, "Market share 100 nahi"
assert (group_share_pct.drop(columns="bank_group").sum() - 100).abs().max() < 1e-6, "Group share 100 nahi"
assert abs(contribution_to_growth["share_of_total_increase_pct"].sum() - 100) < 1e-6, "Contribution 100 nahi"
src_tot = raw[raw["cf_excel"] > 0].groupby("year")["cf_excel"].sum().values
assert np.allclose(market_summary["consumer_finance_bn"].values, src_tot), "Market total source se match nahi"
ties = bank_metrics_all_years[bank_metrics_all_years.duplicated(["year", "rank"], keep=False)]
print("Sab checks pass ✓")
if len(ties): print("Note – barabar CF wale banks (same rank):\n", ties[["year", "bank", "cf_bn", "rank"]].to_string(index=False))
```

## Step 10 – Excel file banayein

```python
read_me = pd.DataFrame({"note": [
    f"Source file: {INPUT_FILE} (sheet '{INPUT_SHEET}'). {cd['bank'].nunique()} banks, {Y0}–{Y1}.",
    "Data = overall Consumer Finance only as given in source; NOT split into personal, auto or housing loans.",
    "Consumer finance (cf_bn) taken directly from source in PKR bn. cf_share = cf_bn / advances (derived in Python).",
    f"CF of 0 in source treated as 'not available' and excluded ({len(has_data) - len(BANKS)} banks have 0 in all years).",
    "Excel review notes are entered manually in the notebook (REVIEW_NOTES); update them if balance-sheet data changes.",
    "Disbursement = change in balance + average balance / tenure; tenures "
        + "/".join(str(v) for v in TENURES.values()) + " years. MODELLED ESTIMATE.",
    "Penetration uses only banks with consumer finance data.",
    "External data (CPI, policy rate) kept separately and labelled External.",
    "See 'issues' sheet for data quality flags to verify against annual reports.",
]})
external_data = pd.DataFrame({"year": range(Y0, Y1 + 1), "cpi_inflation_pct": None,
                              "policy_rate_end_pct": None, "source": None})
fix_decisions = pd.DataFrame(columns=["type", "bank", "year", "value"])

sheets = {
    "read_me": read_me, "market_summary": market_summary, f"bank_metrics_{Y1}": bank_metrics_last,
    "bank_metrics_all_years": bank_metrics_all_years, "concentration": concentration,
    "contribution_to_growth": contribution_to_growth, "group_cf_bn": group_cf_bn,
    "group_share_pct": group_share_pct, "group_penetration_pct": group_penetration_pct,
    "stability": stability, "drb_peers": drb_peers, "external_data": external_data,
    "issues": issues, "fix_decisions": fix_decisions, "clean_data": clean_data,
}

with pd.ExcelWriter(OUTPUT_FILE, engine="openpyxl") as w:
    for name, df in sheets.items():
        df.to_excel(w, sheet_name=name, index=False)
        ws = w.sheets[name]
        for i, col in enumerate(df.columns, start=1):        # column width
            width = max([len(str(col))] + [len(str(v)) for v in df[col].head(200)]) + 2
            ws.column_dimensions[ws.cell(1, i).column_letter].width = min(max(width, 10), 60)
        ws.freeze_panes = "A2"

print("File ban gayi:", OUTPUT_FILE)
```
